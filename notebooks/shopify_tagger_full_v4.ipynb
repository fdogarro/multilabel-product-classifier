## 6. Evaluate Best Checkpoint on Test Set

```python
best_model = BertMultiLabelClassifier.from_checkpoint(CHECKPOINT)
best_model.eval().to(DEVICE)

best_thresholds = results['best_thresholds']

all_probs, all_labels = [], []
with torch.no_grad():
    for batch in test_loader:
        input_ids      = batch['input_ids'].to(DEVICE)
        attention_mask = batch['attention_mask'].to(DEVICE)
        token_type_ids = batch.get('token_type_ids')
        if token_type_ids is not None:
            token_type_ids = token_type_ids.to(DEVICE)

        logits = best_model(input_ids, attention_mask, token_type_ids)
        probs  = torch.sigmoid(logits).cpu().numpy()
        all_probs.append(probs)
        all_labels.append(batch['labels'].numpy())

y_prob = np.vstack(all_probs)
y_true = np.vstack(all_labels)

test_metrics = compute_metrics(y_true, y_prob, best_thresholds)

print('Test set results:')
for k, v in test_metrics.items():
    if k != 'per_label_f1':
        print(f'  {k:<20} {v}')
```

## 8. Inference Latency

```python
# Patch measure_latency to skip cuda.synchronize which corrupts the GPU context
import time, numpy as np

def bert_measure_latency(model, tokenizer, n_samples=100, max_length=120, device=None):
    device = device or ("cuda" if torch.cuda.is_available() else "cpu")
    model.eval().to(device)
    dummy = ["Wireless Bluetooth speaker waterproof outdoor audio device"]
    enc   = tokenizer(dummy, max_length=max_length, padding="max_length",
                      truncation=True, return_tensors="pt")
    ids  = enc["input_ids"].to(device)
    mask = enc["attention_mask"].to(device)
    latencies = []
    with torch.no_grad():
        for _ in range(n_samples):
            start = time.perf_counter()
            model(ids, mask)
            latencies.append((time.perf_counter() - start) * 1000)
    return {
        "mean_ms":   round(float(np.mean(latencies)), 2),
        "median_ms": round(float(np.median(latencies)), 2),
        "p95_ms":    round(float(np.percentile(latencies, 95)), 2),
        "p99_ms":    round(float(np.percentile(latencies, 99)), 2),
    }

print("bert_measure_latency patched — cuda.synchronize removed")
```

## 3. Full Test Set Evaluation — Claude Haiku

```python
def evaluate_llm(provider: str, model: str, df: pd.DataFrame, label_names: list[str]) -> dict:
    """Run the full LLM evaluation loop and return metrics + telemetry."""
    tagger = LLMApiTagger(
        provider=provider,
        model=model,
        taxonomy_path='../data/taxonomy.json',
        cache_dir='../data/raw/llm_cache',
        request_delay=0.3,
    )

    all_preds, all_true = [], []
    latencies = []
    n = len(df)

    for i, row in df.iterrows():
        product = {'title': row['title'], 'description': row['description']}
        start   = time.perf_counter()
        result  = tagger.predict(product)
        latencies.append((time.perf_counter() - start) * 1000)

        all_preds.append(result['labels'])

        # Ground truth
        gt = row['labels']
        if isinstance(gt, str):
            import ast
            gt = ast.literal_eval(gt)
        all_true.append([int(v) for v in gt])

        if (i + 1) % 200 == 0:
            print(f'  {i+1}/{n} complete...')

    y_pred = np.array(all_preds)
    y_true = np.array(all_true)

    # LLM output is already binary — use 0.5 threshold (pass probs = binary pred)
    metrics  = compute_metrics(y_true, y_pred.astype(float), thresholds=[0.5] * len(label_names))
    failures = tagger.failure_report()
    cost     = tagger.cost_estimate(n_products=100_000)

    return {
        'metrics':   metrics,
        'failures':  failures,
        'cost':      cost,
        'latency':   {
            'mean_ms':   round(float(np.mean(latencies)), 2),
            'median_ms': round(float(np.median(latencies)), 2),
            'p95_ms':    round(float(np.percentile(latencies, 95)), 2),
            'p99_ms':    round(float(np.percentile(latencies, 99)), 2),
        },
    }


print('Evaluating Claude Haiku on test set...')
haiku_results = evaluate_llm(
    provider='anthropic',
    model='claude-haiku-4-5-20251001',
    df=test_df,
    label_names=LABEL_NAMES,
)

print('\nClaude Haiku — test set metrics:')
for k, v in haiku_results['metrics'].items():
    if k != 'per_label_f1':
        print(f'  {k:<20} {v}')
```

# Phase 5: Model Selector, Orchestrator & Live Demo

**CSCI E-222 · Spring 2026**

The core contribution of this project: a confidence-calibrated routing layer that assigns each product to the most appropriate model. Results from Phases 2–4 feed directly into the routing policy.

Pipeline flow:
1. **Preprocessor** — tokenize and clean input
2. **BERT inference** — initial confidence estimate
3. **Model Selector** — route based on confidence thresholds
4. **Tagger** — run the selected model
5. **Validator** — check tag coverage and confidence floor

```python
# ── Agent definitions ────────────────────────────────────────────────────────

ROUTE_BERT = "bert"
ROUTE_LORA = "lora_mistral"
ROUTE_LLM  = "llm_api"

class ModelSelector:
    def __init__(self, high_conf_threshold=0.75, low_conf_threshold=0.50):
        self.high_conf = high_conf_threshold
        self.low_conf  = low_conf_threshold

    def select(self, probs):
        max_conf  = float(probs.max())
        mean_conf = float(probs.mean())
        if max_conf >= self.high_conf:
            return ROUTE_BERT, f"BERT high-confidence (max={max_conf:.2f})"
        elif max_conf >= self.low_conf:
            return ROUTE_LORA, f"LoRA-Mistral ambiguous (max={max_conf:.2f}, mean={mean_conf:.2f})"
        else:
            return ROUTE_LLM, f"LLM API fallback (max={max_conf:.2f} < {self.low_conf})"


class ValidatorAgent:
    def __init__(self, min_tags=1, confidence_floor=0.30):
        self.min_tags         = min_tags
        self.confidence_floor = confidence_floor

    def validate(self, probs, tags):
        if len(tags) < self.min_tags:
            return True, f"Too few tags ({len(tags)} < {self.min_tags})"
        if float(max(probs)) < self.confidence_floor:
            return True, f"Low confidence (max={max(probs):.2f} < {self.confidence_floor})"
        return False, None


class Orchestrator:
    def __init__(self, bert_model, tokenizer, selector, validator,
                 label_names, thresholds, device='cpu', max_length=120,
                 lora_model=None, llm_tagger=None):
        self.bert_model  = bert_model
        self.tokenizer   = tokenizer
        self.selector    = selector
        self.validator   = validator
        self.label_names = label_names
        self.thresholds  = thresholds
        self.device      = device
        self.max_length  = max_length
        self.lora_model  = lora_model
        self.llm_tagger  = llm_tagger

    def run(self, product):
        text = f"{product.get('title','')} {product.get('description','')}".strip()
        enc  = self.tokenizer(
            text, max_length=self.max_length,
            truncation=True, padding='max_length', return_tensors='pt'
        )

        # BERT inference for confidence estimate
        self.bert_model.eval()
        with torch.no_grad():
            ids  = enc['input_ids'].to(self.device)
            mask = enc['attention_mask'].to(self.device)
            tids = enc.get('token_type_ids')
            if tids is not None:
                tids = tids.to(self.device)
            logits     = self.bert_model(ids, mask, tids)
            bert_probs = torch.sigmoid(logits).cpu().numpy()[0]

        # Route
        route, routed_by = self.selector.select(bert_probs)

        # Tag with selected model
        if route == ROUTE_BERT or (route == ROUTE_LORA and self.lora_model is None):
            probs      = bert_probs
            model_used = ROUTE_BERT
        elif route == ROUTE_LORA:
            self.lora_model.eval()
            with torch.no_grad():
                logits = self.lora_model(ids, mask)
                probs  = torch.sigmoid(logits).cpu().float().numpy()[0]
            model_used = ROUTE_LORA
        else:
            if self.llm_tagger:
                res    = self.llm_tagger.predict(product)
                probs  = np.array(res['labels'], dtype=float)
            else:
                probs  = bert_probs
            model_used = ROUTE_LLM

        tags = [self.label_names[i]
                for i, (p, t) in enumerate(zip(probs, self.thresholds)) if p >= t]

        flagged, flag_reason = self.validator.validate(probs, tags)

        return {
            'tags':        tags,
            'probs':       probs.tolist(),
            'model_used':  model_used,
            'routed_by':   routed_by,
            'flagged':     flagged,
            'flag_reason': flag_reason,
        }

print("Agents defined: ModelSelector, ValidatorAgent, Orchestrator")
```

## 1. Initialize Pipeline

```python
# Phase 5 always runs on CPU — avoids CUDA context corruption from earlier phases
DEMO_DEVICE = 'cpu'

selector  = ModelSelector(high_conf_threshold=0.75, low_conf_threshold=0.50)
validator = ValidatorAgent(min_tags=1, confidence_floor=0.30)

bert_ckpt = f'../checkpoints/{MODEL_NAME.replace("/", "-")}'

# Verify checkpoint before loading
with open(f'{bert_ckpt}/config.json') as f:
    ckpt_config = json.load(f)
print('Checkpoint config:', ckpt_config)

# Load on CPU and keep there — CUDA context may be corrupt after Phases 2–4
pipeline_model = BertMultiLabelClassifier.from_checkpoint(bert_ckpt)  # loads to cpu by default
pipeline_model = pipeline_model.cpu()  # explicit .cpu() to guarantee placement
pipeline_model.eval()

# Verify all parameters are on CPU before building orchestrator
for pname, param in pipeline_model.named_parameters():
    assert param.device.type == 'cpu', f'{pname} is still on {param.device}'
print(f'Pipeline model on: {next(pipeline_model.parameters()).device}')

pipeline_tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

orchestrator = Orchestrator(
    bert_model  = pipeline_model,
    tokenizer   = pipeline_tokenizer,
    selector    = selector,
    validator   = validator,
    label_names = LABEL_NAMES,
    thresholds  = best_thresholds,
    device      = DEMO_DEVICE,
    lora_model  = None,
    llm_tagger  = tagger,
)
print('Pipeline ready on CPU.')
```

```python
import gradio as gr

def predict_product(title, description):
    if not title.strip() and not description.strip():
        return {"(no input)": 0.0}, {"error": "Please enter a title or description."}
    product = {'title': title, 'description': description}
    result  = orchestrator.run(product)
    tag_scores = {tag: float(result['probs'][LABEL_NAMES.index(tag)])
                  for tag in result['tags']} if result['tags'] else {"(no tags predicted)": 0.0}
    routing = {
        "model_used":  result["model_used"],
        "routed_by":   result["routed_by"],
        "flagged":     str(result["flagged"]),
        "flag_reason": result["flag_reason"] or "none",
        "n_tags":      len(result["tags"]),
    }
    return tag_scores, routing

demo = gr.Interface(
    fn=predict_product,
    inputs=[
        gr.Textbox(label="Product Title",
                   placeholder="e.g. Sony WH-1000XM5 Wireless Headphones"),
        gr.Textbox(label="Product Description", lines=4,
                   placeholder="e.g. Industry-leading noise cancellation, 30-hour battery..."),
    ],
    outputs=[
        gr.Label(label="Predicted Tags", num_top_classes=10),
        gr.JSON(label="Routing Decision"),
    ],
    title="Shopify Product Tagger — Multi-Agent Pipeline",
    description=(
        "Enter a product title and description. "
        "The pipeline routes to BERT (high confidence), LoRA-Mistral (ambiguous), "
        "or LLM API (out-of-distribution) based on confidence scores."
    ),
    examples=[
        ["Sony WH-1000XM5 Wireless Headphones",
         "Industry-leading noise cancellation, 30-hour battery, multipoint connection."],
        ["Cast Iron Skillet 12-inch",
         "Pre-seasoned, oven safe to 500°F, ideal for searing, baking, and frying."],
        ["Kids Dinosaur Backpack",
         "Cute T-Rex design, multiple pockets, padded straps, fits kids ages 3-8."],
        ["Organic Green Tea 100 bags",
         "USDA certified organic, antioxidant-rich, individually wrapped tea bags."],
    ],
)

demo.launch(share=True, debug=True)  # share=True gives a public URL in Colab
```
